```python
Q1=data['petal.length'].quantile(0.25)
Q3=data['petal.length'].quantile(0.75)
Q1,Q3
```

```python
IQR=Q3-Q1
IQR
```

```python
lower_limit=Q1-1.5*IQR
Upper_limit=Q3+1.5*IQR
lower_limit,Upper_limit
```

```python
data_which_are_outlier=data[(data['petal.length']<lower_limit) | (data['petal.length']>Upper_limit)]
data_which_are_outlier
```

```python
data_which_areNOT_outlier=data[(data['petal.length']>lower_limit) & (data['petal.length']<Upper_limit)]
data_which_areNOT_outlier
```

```python
#here we are seeing outlier in sepal.width
#removing outlier and storing in new dataset
Q1=data['sepal.width'].quantile(0.25)
Q3=data['sepal.width'].quantile(0.75)
IQR=Q3-Q1
lower_limit=Q1-1.5*IQR
upper_limit=Q3+1.5*IQR
dataset_with_NoOutlier=data[(data['sepal.width']>lower_limit) & (data['sepal.width']<upper_limit)]
dataset_with_NoOutlier
```

```python
dataset_with_NoOutlier.boxplot()
```

```python
plt.title("Plant Variety")
sns.countplot(dataset_with_NoOutlier['variety'])
plt.show()
for col in dataset_with_NoOutlier.select_dtypes(include=['float64','object']).columns:
    print(dataset_with_NoOutlier[col].value_counts())
```
