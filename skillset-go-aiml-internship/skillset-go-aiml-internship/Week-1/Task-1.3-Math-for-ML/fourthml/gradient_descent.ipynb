```python
import numpy as np
```

```python
def gradient(x,y):
    n=len(x)
    b=m=0
    iteration=1000
    a=0.0001
    for i in range(iteration):
        ypred= (m*x) + b
        dm = (-2/n)*(sum((y-ypred)*x))
        db = (-2/n)*(sum((y-ypred)))
        m = m - (a*dm)
        b = b - (a*db)
        cost= (1/n)*sum(val**2 for val in (y-ypred))
        print(f" the value of slope m is : {m} \n and the value of y intercept b is : {b} \n at iteration : {i} \n value of y_after_iteration: {ypred} \n the cost in this iteration is : {cost} " )
        print('\n')
x=np.array([34,53,22,11,5])
y=np.array([45,33,21,66,2])
#[33.15140919 51.56181726 21.52378304 10.86512573  5.05131266]
gradient(x,y)
```
